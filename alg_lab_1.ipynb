Шейкерная сортировка

```python
from random import randint

lenght = 20
arr = [randint(1, 100) for i in range(1, lenght)]

print(arr)


def shakerSort(arr):
    lenght = len(arr)
    swapped = True
    el = 0
    end = lenght - 1

    while (swapped == True):

        swapped = False

        for i in range(el, end):
            if (arr[i] > arr[i+1]):
                arr[i], arr[i+1] = arr[i+1], arr[i]
                swapped = True

        if (not (swapped)):
            break

        swapped = False

        end -= 1

        for i in range(end-1, el-1, -1):
            if (arr[i] > arr[i+1]):
                arr[i], arr[i+1] = arr[i+1], arr[i]

                swapped = True
        el += 1


shakerSort(arr)
print(arr)
```

Сортировка Шелла

```python
from random import randint

lenght = 20
arr = [randint(1, 100) for i in range(1, lenght)]

print(arr)


def shellSort(arr):
    end = len(arr)
    step = len(arr)//2

    while (step > 0):
        for i in range(step, end):
            el = i - step
            while el >= 0 and arr[el] > arr[i]:

                arr[el], arr[i] = arr[i], arr[el]

                i = el
                el = i - step
        step //= 2


shellSort(arr)
print(arr)
```

Поиск по бору

```python
class TrieNode:
    def __init__(self):
        self.children = {}
        self.end = False
        self.index = -1

    def getChild(self):
        return self.children
```

```python
class Trie:
    def __init__(self):
        self.root = TrieNode()

    def getRoot(self):
        return self.root

    def insert(self, word, index):
        node = self.root
        for char in word:
            node = node.children.setdefault(char, TrieNode())
        node.end = True
        node.index = index

    def search(self, word):
        node = self._find_node(word)
        if node and node.end:
            return node.index
        return -1

    def _find_node(self, prefix):
        node = self.root
        for char in prefix:
            if char not in node.children:
                return None
            node = node.children[char]
        return node
```

```python
trie = Trie()
words = ["hello", "herra", "hell", "hero", "her", "hellicopter"]
for i, word in enumerate(words):
    trie.insert(word, i)
print(trie.getRoot().getChild())
print(trie.search("herra"))
```

## 1. Шейкерная сортировка (Shaker Sort / Cocktail Shaker Sort)

### Описание
Шейкерная сортировка — это улучшенная версия пузырьковой сортировки. Она обходит список в обоих направлениях (сначала слева направо, затем справа налево), "встряхивая" элементы к их правильным позициям. Каждый проход перемещает следующий наименьший элемент в начало и следующий наибольший элемент в конец.

### Реализация
В ноутбуке представлена функция `shakerSort(arr)`, которая реализует данный алгоритм.

### Результаты выполнения
Исходный массив случайных чисел сортируется, и выводится отсортированный массив.
Пример:
*   Исходный массив: `[28, 29, 11, 61, 1, 33, 69, 73, 41, 69, 98, 24, 35, 79, 81, 13, 16, 93, 35]`
*   Отсортированный массив: `[1, 11, 13, 16, 24, 28, 29, 33, 35, 35, 41, 61, 69, 69, 73, 79, 81, 93, 98]`

## 2. Сортировка Шелла (Shell Sort)

### Описание
Сортировка Шелла — это обобщение сортировки вставками. Она работает, сравнивая элементы, находящиеся на некотором расстоянии друг от друга, а затем постепенно уменьшает это расстояние (шаг). Благодаря этому элементы могут перемещаться на большие расстояния за один проход, чем при обычной сортировке вставками.

### Реализация
В ноутбуке представлена функция `shellSort(arr)`, которая реализует данный алгоритм, используя уменьшающийся шаг (`step = len(arr)//2`).

### Результаты выполнения
Исходный массив случайных чисел сортируется, и выводится отсортированный массив.
Пример:
*   Исходный массив: `[89, 18, 94, 32, 55, 88, 78, 78, 6, 33, 32, 20, 15, 54, 78, 46, 19, 53, 19]`
*   Отсортированный массив: `[6, 15, 18, 19, 19, 20, 32, 32, 33, 46, 53, 54, 55, 78, 78, 78, 88, 89, 94]`

## 3. Поиск по бору (Trie / Prefix Tree)

### Описание
Бор (Trie) — это древовидная структура данных, используемая для эффективного хранения и извлечения строк (или последовательностей). Каждый узел в боре представляет собой префикс, а дочерние узлы представляют следующие символы в строке. Это позволяет быстро искать слова по префиксу.

### Реализация
*   **`TrieNode`**: Класс представляет узел бора, содержащий словарь `children` для дочерних узлов, флаг `end` (указывает, является ли этот узел концом слова) и `index` (для хранения индекса слова).
*   **`Trie`**: Класс реализует функциональность бора, включая:
    *   `__init__`: Инициализирует корень бора.
    *   `insert(word, index)`: Вставляет слово в бор, помечая конечный узел и сохраняя индекс слова.
    *   `search(word)`: Ищет слово в боре и возвращает его индекс, если найдено.

### Результаты выполнения
Пример вставки слов `["hello", "herra", "hell", "hero", "her", "hellicopter"]` в бор и последующего поиска слова "herra".
*   `trie.getRoot().getChild()` показывает структуру дочерних узлов от корня.
*   `trie.search("herra")` возвращает `1`, что соответствует индексу слова "herra" в исходном списке `words`.
