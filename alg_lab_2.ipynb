```python
def prim(n, edges):
    graph = [[] for _ in range(n)]
    for u, v, w in edges:
        graph[u].append((w, v))
        graph[v].append((w, u))

    visited = [False] * n
    min_edges = [(0, 0)]
    total_weight = 0

    while min_edges:
        min_edge = min(min_edges, key=lambda x: x[0])
        min_edges.remove(min_edge)
        w, u = min_edge

        if visited[u]:
            continue
        visited[u] = True
        total_weight += w

        for nw, v in graph[u]:
            if not visited[v]:
                min_edges.append((nw, v))

    return total_weight

edges = [
    (0, 1, 4), (0, 2, 4), (1, 2, 2), (1, 3, 5),
    (2, 3, 8), (2, 4, 10), (3, 4, 2), (3, 5, 6), (4, 5, 3)
]

print("Общий вес МОД:", prim(6, edges))
```

# Лабораторный отчет: Реализация и сравнение алгоритмов Краскала и Прима

## Введение

Данный отчет представляет собой анализ и сравнение двух известных алгоритмов для нахождения минимального остовного дерева (МОД) во взвешенном неориентированном графе: алгоритмов Краскала и Прима. Оба алгоритма являются фундаментальными в теории графов и имеют широкое применение в различных областях, таких как проектирование сетей, телекоммуникации и кластеризация данных.

## Описание алгоритмов

### Алгоритм Краскала

Алгоритм Краскала строит минимальное остовное дерево, последовательно добавляя ребра с наименьшим весом, если они не образуют циклы с уже выбранными ребрами. Основная идея заключается в использовании структуры данных Disjoint Set Union (DSU) для эффективного отслеживания компонентов связности графа. На каждом шаге выбирается следующее наименьшее ребро; если его концы находятся в разных компонентах связности, ребро добавляется к МОД, и компоненты объединяются. Этот процесс продолжается до тех пор, пока не будет добавлено n-1 ребро, где n - количество вершин.

### Алгоритм Прима

Алгоритм Прима строит минимальное остовное дерево, начиная с произвольной вершины и постепенно расширяя дерево. На каждом шаге выбирается ребро с минимальным весом, которое соединяет вершину из уже построенной части МОД с вершиной, еще не включенной в дерево. Для эффективного выбора такого ребра часто используется приоритетная очередь. Алгоритм продолжает работу до тех пор, пока все вершины графа не будут включены в МОД.

## Реализация

Реализации обоих алгоритмов были проведены с использованием языка программирования Python. Для алгоритма Краскала был использован подход с сортировкой всех ребер и применением DSU для проверки на циклы. Для алгоритма Прима была использована приоритетная очередь (представленная списком, имитирующим ее поведение), которая хранит ребра, соединяющие посещенные и непосещенные вершины.

Вспомогательные функции:

*   **`encodeGraph(graph)`**: Функция для кодирования входного списка ребер в более удобный формат, представляющий граф как словарь смежности или список, где ключом является пара вершин, а значением - вес ребра.
*   **`sortGraph(graph)`**: Функция для сортировки графа по весам ребер, что критически важно для эффективной работы алгоритма Краскала.
*   **`findVert(graph)`**: Функция для нахождения уникальных вершин в графе, полезная для инициализации структур данных.

## Сравнение производительности и применения

### Сложность

*   **Алгоритм Краскала**: Временная сложность обычно составляет O(E log E) или O(E log V) (где E - количество ребер, V - количество вершин) из-за сортировки ребер. Операции DSU (find и union) имеют почти константную амортизированную сложность.
*   **Алгоритм Прима**: Временная сложность зависит от используемой структуры данных для приоритетной очереди. С использованием бинарной кучи сложность составляет O(E log V), а с использованием фибоначчиевой кучи - O(E + V log V).

### Применение

*   **Алгоритм Краскала** лучше подходит для разреженных графов (с небольшим количеством ребер), так как его производительность в значительной степени зависит от количества ребер.
*   **Алгоритм Прима** более эффективен для плотных графов (с большим количеством ребер), поскольку его производительность больше зависит от количества вершин.

В общем случае, оба алгоритма дают корректный результат - находят минимальное остовное дерево. Выбор между ними часто определяется структурой входного графа и конкретными требованиями к производительности.
