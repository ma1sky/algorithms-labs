Task #1

````python
import math

edges = {
    (1, 2): 9, (1, 3): 2,
    (2, 4): 4, (2, 5): 12, (3, 2): 6,
    (3, 5): 28, (3, 6): 10,
    (4, 5): 16, (6, 5): 15,
    (5, 7): 3, (5, 6): 15,
    (6, 7): 25, (4, 7): 19
}
````

````python
def dynamic_programming(graph, start, end):
    distances = {}
    for i in range(1, 8):
        distances[i] = math.inf
    distances[start] = 0

    predecessors = {}
    for i in range(1, 8):
        predecessors[i] = None

    for _ in range(len(graph)):
        for (u, v), weight in graph.items():
            if distances[u] + weight < distances[v]:
                distances[v] = distances[u] + weight
                predecessors[v] = u

    shortest_distance = distances[end]
    path = []
    current_vertex = end

    if shortest_distance == float('inf'):
        return None, []

    while current_vertex is not None:
        path.append(current_vertex)
        current_vertex = predecessors[current_vertex]

    path.reverse()
    return shortest_distance, path
````

````python
shortest_distance, path = dynamic_programming(edges, 1, 7)
print("Кратчайшее расстояние:", shortest_distance)
print("Кратчайший путь:", path)
````

Task #2

````python
graph = {
    1: [(2, 8), (4, 5), (6, 9)],
    2: [(1, 8), (3, 2), (4, 1)],
    3: [(2, 2), (5, 3), (8, 10)],
    4: [(1, 5), (2, 1), (5, 9), (6, 3)],
    5: [(3, 3), (4, 9), (7, 12), (8, 6)],
    6: [(1, 9), (4, 3), (7, 12)],
    7: [(5, 12), (6, 12), (8, 7)],
    8: [(3, 10), (5, 6), (7, 7)]
}
````

````python
def dijkstra(graph, start, end):
    distances = {}
    for i in range(1, 8):
        distances[i] = math.inf
    distances[start] = 0

    previous = {}
    for node in graph:
        previous[node] = None

    visited = set()

    while len(visited) < len(graph):
        unvisited_nodes = []
        for node in distances:
            if node not in visited:
                unvisited_nodes.append(node)

        current = None
        min_distance = float('inf')

        for node in unvisited_nodes:
            if distances[node] < min_distance:
                min_distance = distances[node]
                current = node
                if current == end:
                    break

        visited.add(current)

        for neighbor, weight in graph[current]:
            new_distance = distances[current] + weight
            if new_distance < distances[neighbor]:
                distances[neighbor] = new_distance
                previous[neighbor] = current

    path = []
    while end is not None:
        path.append(end)
        end = previous[end]
    path.reverse()

    return distances[path[-1]], path
````

````python
shortest_distance, shortest_path = dijkstra(graph, 1, 8)
print("Кратчайшее расстояние:", shortest_distance)
print("Кратчайший путь:", shortest_path)
````
