````python
import math

def tsp_branch_and_bound(graph):
    n = len(graph)
    visited = [False] * n
    best_cost = math.inf
    best_path = []

    def bound(current_cost, level):
        min_edges = []
        for row_idx in range(n):
            if not visited[row_idx]:
                min_edge = min((graph[row_idx][i] for i in range(n) if not visited[i] and graph[row_idx][i] != math.inf), default=0)
                min_edges.append(min_edge)
            else:
                min_edges.append(0)
        return current_cost + sum(min_edges[:n - level])

    def solve(current_node, level, current_path, current_cost):
        nonlocal best_cost, best_path

        if level == n:
            total_cost = current_cost + graph[current_node][0]
            if total_cost < best_cost:
                best_cost = total_cost
                best_path = current_path + [0]
            return

        for next_node in range(n):
            if not visited[next_node] and graph[current_node][next_node] != math.inf:
                next_cost = current_cost + graph[current_node][next_node]
                if bound(next_cost, level) < best_cost:
                    visited[next_node] = True
                    solve(next_node, level + 1, current_path + [next_node], next_cost)
                    visited[next_node] = False

    visited[0] = True
    solve(0, 1, [0], 0)

    return best_cost, best_path

distance_matrix = [
    [math.inf, 7, 12, 25, 10],
    [10, math.inf, 9, 5, 11],
    [13, 8, math.inf, 6, 4],
    [6, 11, 15, math.inf, 15],
    [5, 9, 12, 17, math.inf]
]

best_cost, best_path = tsp_branch_and_bound(distance_matrix)
print("Минимальная стоимость:", best_cost)
print("Оптимальный путь:", best_path)
````
